```python
import pandas as pd
from pathlib import Path

data_dir = Path("../data")

files = sorted(data_dir.glob("yellow_tripdata_*.parquet"))

frames = []

for parquet_file in files:
    # Lire le fichier
    df = pd.read_parquet(parquet_file)

    # Extraire le mois depuis le nom du fichier
    month = parquet_file.stem.replace("yellow_tripdata_", "")

    # Ajouter une colonne source_month
    df["source_month"] = month

    # Ajouter le DataFrame à la liste
    frames.append(df)

# Fusionner tous les DataFrames
trips = pd.concat(frames, ignore_index=True)

# Vérifier le résultat
print(trips.shape)

trips.groupby("source_month").size()
```
